Start by just choosing the first available move (will always be keep in this situation)

```python
while not env.game_over:
    player = env.current_player
    legal_actions = env.legal_actions()

    print(f"\nPlayer: {player}")
    print(f"Legal actions: {legal_actions}")
    print(f"Drawn card: {obs['drawn_card']}")

    # Start with the first legal action
    action = legal_actions[0]
    print(f"Selected: {action_name(action)}")

    obs, rewards, done, info = env.step(action)

print("\nGame finished!")
print("Scores:", env.scores)
print("Winner:", env.winner)
```

Now switch to random selection of legal moves

Mapping for the actions

| Action Number | Action Name | Description |
|---|---|---|
| 0 | KEEP | Keep the drawn card |
| 1 | PEEK | Look at one of your unknown cards |
| 2 | TRADE(0) | Trade with opponent's card at position 0 |
| 3 | TRADE(1) | Trade with opponent's card at position 1 |
| 4 | TRADE(2) | Trade with opponent's card at position 2 |
| 5 | TRADE(3) | Trade with opponent's card at position 3 |
| 6 | CALL | Call to end the game |

```python
env = HiddenHandEnv(seed=10)
obs = env.reset()

rng = rand.Random(42)

while not env.game_over:

    player = env.current_player
    legal_actions = env.legal_actions()

    print(f"\nPlayer: {player}")
    print(f"Legal actions: {legal_actions}")
    print(f"Drawn card: {obs['drawn_card']}")

    action = rng.choice(legal_actions)

    print(f"Selected: {action_name(action)}")

    obs, rewards, done, info = env.step(action)

print("\nFinal scores:", env.scores)
print("Winner:", env.winner)
```

Comparing Observations Before and After an Action

```python
env = HiddenHandEnv(seed=10)
obs_before = env.reset()

player_before = env.current_player

print("BEFORE ACTION")
print("Player:", player_before)
print("Board:", obs_before["board_views"])
print("Drawn card:", obs_before["drawn_card"])

# Choose an action
action = env.legal_actions()[0]

print("\nSelected:", action_name(action))

# Execute the action
obs_after, rewards, done, info = env.step(action)

print("\nAFTER ACTION")
print("Player:", env.current_player)
print("Board:", obs_after["board_views"])
print("Drawn card:", obs_after["drawn_card"])
print("Rewards:", rewards)
```
