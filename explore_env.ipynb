Now switch to random selection of legal moves

Mapping for the actions

| Action Number | Action Name | Description |
|---|---|---|
| 0 | KEEP | Keep the drawn card |
| 1 | PEEK | Look at one of your unknown cards |
| 2 | TRADE(0) | Trade with opponent's card at position 0 |
| 3 | TRADE(1) | Trade with opponent's card at position 1 |
| 4 | TRADE(2) | Trade with opponent's card at position 2 |
| 5 | TRADE(3) | Trade with opponent's card at position 3 |
| 6 | CALL | Call to end the game |

````python
env = HiddenHandEnv(seed=10)
obs = env.reset()

rng = rand.Random(42)

while not env.game_over:

    player = env.current_player
    legal_actions = env.legal_actions()

    print(f"\nPlayer: {player}")
    print(f"Legal actions: {legal_actions}")
    print(f"Drawn card: {obs['drawn_card']}")

    action = rng.choice(legal_actions)

    print(f"Selected: {action_name(action)}")

    obs, rewards, done, info = env.step(action)

print("\nFinal scores:", env.scores)
print("Winner:", env.winner)
````

## Exploring Observation Tokens

Different kinds of tokens hidden hand environment provides

- **board_tokens:** Information about the cards on the board.
- **discard_token:** Information about discarded cards.
- **global_token:** General information about the game.
- **decision_token:** Information about the current decision.
- **action_tokens:** Information about previous actions.

Understand the tokens so we can convert them into numerical tensors that MuZero can represent in its networks.

Inspect each kind of token and their values

````python
env = HiddenHandEnv(seed=10)
obs = env.reset()

tokens = obs["tokens"]

for name, values in tokens.items():
    print(f"\n{name}:")
    print(values)
````

***Note that this information can also be found at the top of `env.py` and in the `observe()` method.***

## Examining the Tokens

### Board Tokens

From our example:

```python
board_tokens = [
    [1, 0, 0, -99],
    [1, 1, 1, 3],
    [1, 2, 0, -99],
    [1, 3, 0, -99],
    [0, 0, 0, -99],
    [0, 1, 0, -99],
    [0, 2, 0, -99],
    [0, 3, 0, -99]
]
```

Each element represents:

`[owner, position, is_known, card_value]`

For example, consider `[1, 1, 1, 3]`:

- The card belongs to player 1.
- The card is in position 1.
- The card's value is known to the observing player.
- The card's value is 3.

The value `-99` indicates that the card's value is unknown to the observing player.

**Key Takeaway:** We have 8 cards with 4 features each.

Therefore, the board dimensions are **8 × 4**, containing 32 numerical entries.

---

### Discard Tokens

From our example:

```python
discard_token = [-99, 0]
```

Each element represents:

`[discard_top, discard_pile_size]`

For example, consider `[-99, 0]`:

- `-99`: No card is currently on top of the discard pile because the pile is empty.
- `0`: The discard pile contains 0 cards.

If our token were `[7, 2]`, this would mean the top card has a value of 7 and there are 2 cards in the discard pile.

**Key Takeaway:** The discard token contains **2 features**, giving it a dimension of **2**. It provides information about the most recently discarded card and the number of discarded cards.

---

### Global Tokens

From our example:

```python
global_token = [19, 0, 1, 0, 0, 0]
```

Each element represents:

`[deck_size, step, current_player, phase, phase_bucket, round_index]`

For example, consider `[19, 0, 1, 0, 0, 0]`:

- `19`: There are 19 cards remaining in the draw pile.
- `0`: No player actions have been taken yet.
- `1`: It is currently player 1's turn.
- `0`: The game is currently in the PLAY phase.
- `0`: The game is in its early stage.
- `0`: The round index is 0.

There are three possible game phases:

| Value | Phase | Description |
|---|---|---|
| 0 | PLAY | The game is proceeding normally. |
| 1 | FINAL_TURN | The game is in its final-turn phase. |
| 2 | GAME_OVER | The game has finished. |

The `phase_bucket` divides the game's progress into three categories:

| Value | Stage | Progress |
|---|---|---|
| 0 | Early | Less than 34% |
| 1 | Middle | 34% to less than 67% |
| 2 | Late | 67% or greater |

The `round_index` is currently always 0 in this implementation.

**Key Takeaway:** The global token contains **6 features**, giving it a dimension of **6**. It provides information about the overall game state, such as whose turn it is and how far the game has progressed.

---

### Action Tokens

From our example:

```python
action_tokens = [
    [0, 2, 0, 0, 1, -127],
    [1, 2, 0, 1, 0, -127]
]
```

Each element represents:

`[actor, action_type, source, target_pos, a_value, b_value]`

The features represent:

- `actor`: The player who performed the action.
- `action_type`: The type of action performed.
- `source`: The acting player's card position used in a trade (0 for other action types).
- `target_pos`: The position of the card involved in the action.
- `a_value`: The public HIGH/LOW tell associated with the action, when applicable.
- `b_value`: The value of the discarded card, when applicable.

The action types are represented by:

| Value | Action Type |
|---|---|
| 1 | KEEP |
| 2 | PEEK |
| 3 | TRADE |
| 4 | CALL |

Note that these are the **public history action types**, which differ from the action IDs used to select moves.

For example, consider:

```python
[0, 2, 0, 0, 1, -127]
```

This means:

- `0`: Player 0 performed the action.
- `2`: The action was PEEK.
- `0`: The source field is 0 (unused for PEEK).
- `0`: The player looked at the card in position 0.
- `1`: The public tell was HIGH.
- `-127`: No discarded card value was recorded for this action.

The values for public tells are:

- `0`: LOW
- `1`: HIGH
- `-127`: No tell or recorded value is available.

Unlike the board tokens, the number of action tokens can change throughout the game.

The environment stores up to the most recent **20 public actions** by default.

**Key Takeaway:** Each action token contains **6 features**, but the number of tokens is variable. This provides historical information about player behavior, which may help MuZero reason about hidden cards.

---

### Decision Tokens

From our example:

```python
decision_token = [1, 7]
```

Each element represents:

`[decision_phase, drawn_value]`

For example, consider `[1, 7]`:

- `1`: The observing player is the current player and can see the drawn card.
- `7`: The currently drawn card has a value of 7.

The first feature acts as a flag:

- `1`: The observing player can see the drawn card.
- `0`: The observing player cannot see the drawn card.

When the drawn card is not visible, its value is represented by `-99`.

For example:

```python
decision_token = [0, -99]
```

This indicates that the drawn card is not visible to the observing player.

**Key Takeaway:** The decision token contains **2 features**, giving it a dimension of **2**. It provides information about the current player's drawn card and whether that information is visible.

---

## Overall Takeaways

After examining all five token types, we can summarize their structures:

| Token Type | Dimensions | Purpose |
|---|---|---|
| Board Tokens | 8 × 4 | Represents the cards and their visibility |
| Discard Token | 2 | Represents the top discarded card and discard pile size |
| Global Token | 6 | Represents the overall game state |
| Action Tokens | Variable × 6 | Represents previous public actions |
| Decision Token | 2 | Represents the current decision and drawn card |
